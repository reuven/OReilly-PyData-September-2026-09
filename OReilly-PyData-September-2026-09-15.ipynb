Data science is really a whole bunch of new disciplines, all related:

- Data analytics -- basically asks to understand what we've already seen

# Slices with `.loc` and `.iloc` are confusing!

`.iloc` uses the traditional way that Python retrieves things, including with slices. This means that the endpoint is "up to and not including." So when we say `.iloc[2:4]`, this means that we want starting with index 2, up to and not including index 4.

However, `.loc` works with whatever index we set, and that index might well not be integers, and might have repeated values, and might be strings. What does it mean to say "up to and not including" in such contexts? That's unclear, so `.loc` just returns up to AND INCLUDING when we give it a slice.

```python
s.describe()  # this returns a series!
```

```python
s.describe().iloc[2]   # get the value for std, based on its position
```

```python
s.describe().loc['std']  # get the value for std, based on the index!
```

# Method chaining

So far, we're writing fairly simple queries. But they do sometimes have multiple invocations of methods. We can, if we want, keep these all on the same line. But that can get very long and unreadable pretty quickly.

```python
# calculate the mean of the first 5 elements of highs
highs.loc[:4].mean()
```

Method chaining is a style of coding in Python that depends on the fact that Python is very strict about whitespace, newlines, and indentation *EXCEPT* when the code is inside of parentheses. When that happens, a lof of the rules get relaxed.

What we do is open `(` -- not because we have to, but because we want to, in order to benefit -- and then we put one method call per line,.

```python
(
    highs       # retrieve the series
    .loc[:4]    # get the first 5 elements -- yes, 5, since it's .loc
    .mean()     # calculate the mean
)
```

# Mask/boolean indexes

Let's review and then combine two ideas we've seen, plus 1-2 new ideas.

1. Fancy indexing -- we know that if we put multiple indexes into a list, and give that to `.loc[]`, we'll get a number of values back.
2. Broadcasting -- we know that if we apply an operator to a series and a scalar value, we get back a new series, one in which the original series' elements have applied that operation.

```python
# what if we broadcast not an arithmetic operation, but rather a boolean/location operation?

s = Series([10, 20, 30, 40, 50])

s == 30    
```

```python
# we've seen fancy indexing with integers (or with other index values). But what if we use fancy indexing with booleans?
# what is -- put a list of True/False values inside of `.loc[]` -- what happens?

s.loc[ [False, False, True, False, False] ]
```

If we do fancy indexing with booleans, then we filter the series -- we only get those elements whose positions correspond to `True` values. Anything with a `False` value is ignored.

We can create such a list of booleans via broadcasting. 

This is how we filter data in Pandas:

1. We perform a comparison using broadcasting. We get back a boolean series.
2. We apply this boolean series to our series as an index, a "mask index."
3. This returns a new series -- the values of the original one, but only where there was a `True`.

```python
# I only want the value 30 from the series

# inside of the square brackets is our broadcast comparison, returning a boolean series
# outside of the square brackets is s.loc, which will filter s

s.loc[ s == 30 ]
```

```python
# I only want values that are >= 30

# inside of the square brackets is our broadcast comparison, returning a boolean series
# outside of the square brackets is s.loc, which will filter s

s.loc[ s >= 30 ]
```

```python
# I only want values that are >= the mean

# inside of the square brackets is our broadcast comparison, returning a boolean series
# outside of the square brackets is s.loc, which will filter s

s.loc[ s >= s.mean() ]
```

# Order of reading/executing mask index

1. Inside of the square brackets, we create a boolean series -- either comparing the series' values with a number or the result of a calculation. The boolean series we create will be our mask index.
2. We apply the mask index using `.loc`.

```python
# Rebecca

s = Series([10, 20, 30, 40, 50])

# s == 30

s.loc[ [False, False, True, False, False] ]   # mask index: True means, "I want the corresponding element," and False means, "I don't."
```

```python
s.loc[ [True, False, True, False, True] ]
```

```python
s.loc[ s==30 ]  # here, s==30 returns a boolean series, which .loc uses as a mask index
```

```python
# what if I want the sum of the even numbers?

s = Series([10, 15, 20, 25, 30, 35, 40, 45, 47, 50])

# 1. Find all of the even numbers by creating a mask index!
# 2. Sum them
```

```python
# we can find even numbers using %2 -- we divide by 2 and check if the remainder is 0. If so, it's even!

s.loc[ s % 2 == 0 ].sum()
```

# dtypes

We've already seen that every series has a dtype, meaning the underlying C data structure that is used to store the information in your series. All of the values in a series must have the same dtype, meaning that they all must be of the same type.

However, dtypes are *not* the same as Python types! In Python, we have `int`, which is all we need for any integer, no matter how big or how small. Not so with dtypes -- we have 64-bit integers, 64-bit unsigned integers, 32-bit integers, 32-bit unsigned integers, etc. The reason we need to start worrying about the size of our values is that they'll be stored in a C-style array, which has C-style values. These are far more compact and more efficient than anything Python itself can offer. But we have to think about them.

Why not just use the largest ones, e.g., `int64`, for everything? Not every integer is going to need 64 bits -- e.g., ages and shoe sizes. Having something that's unnecessarily big wastes memory. Something that's too small risks corrupting your data.

# Default dtypes

If we don't specify what dtype we want, Pandas chooses for us:

- If our series contains only integers, we get `int64`
- If our series contains only numbers, and that includes floats, we get `float64`
- If our series contains some text, we get `str` (as of Pandas 3)
- Otherwise, we get `object`, meaning "any old Python object that I don't know what to do with."

We can specify the dtype when we create a series with the `dtype` keyword argument.

```python
s = Series([10, 20, 30, 40, 50])
s.dtype  # what is its dtype?
```

```python
s = Series([10, 20, 30, 40, 50], dtype='int32')
s.dtype  # what is its dtype?
```

```python
s = Series([10, 20, 30, 40, 50], dtype='int8')   # 8 bits means we have from -128 to 127
s.dtype  # what is its dtype?
```

```python
# what about broadcasting?

s + 100
```

# NaN -- "not a number"

`NaN` is used as a placeholder for corrupt data or data that will come in near future (we hope), or data that is missing. To assume a value instead of `NaN` is often a big mistake.
