# Jupyter crash course

We type into *cells* in Jupyter. Each cell can contain Python (code) or Markdown (documentation, an easier way to write HTML). Each cell has two modes:

- Edit mode: Whenever I type something, the character(s) go into the cell. You can enter edit mode by pressing ENTER or by clicking inside of the cell. (I'm currently in edit mode.)
- Command mode: Whenever I type something, the SINGLE character is taken as a Jupyter command, and affects what Jupyter does. To get into command mode, you can press ESC or click to left of a cell.

What commands are available in command mode?

- `c` -- copy a cell
- `x` -- cut a cell
- `v` -- paste the most recently cut/copied cell
- `a` -- adds a new cell *above* the current one
- `b` -- adds a new cell *below* the current one
- `y` -- makes this cell a *Python* code cell
- `m` -- makes this cell a *Markdown* documentation cell.

When you're done writing in a cell, or if you've edited it and want to re-run it, just use shift+ENTER together. (You can also press the "play" button at the top of the screen.)

# Why Python for data?

Data science is really a whole bunch of new disciplines, all related:

- Data analytics -- basically asks to understand what we've already seen
- Data engineering -- how do we get data from one place to another, quickly and easily
- Machine learning -- asks to make predictions based on existing data

In all of these cases, we're working with lots of data -- millions and billions of numbers.

So... how can it be that Python is the #1 programming language for data? That makes no sense! Python uses a lot of memory, isn't very efficient in its memory usage, and isn't very efficient in its execution speed.

The secret is NumPy, which was written more than 20 years ago, and basically gives a thin Python wrapper around C code for storing and manipulating data. Most data stuff in Python uses NumPy somewhere, and thus benefits from the speed and efficiency of C, while keeping the friendliness of Python.
