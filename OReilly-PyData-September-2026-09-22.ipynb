# Data frames

If a series is a 1D data structure (with values and an index), then a data frame is a 2D data structure, much like an Excel spreadsheet. It also has an index (identifying the rows) and columns (identifying the ... columns).

Each column in a data frame is actually a series in disguise!

```python
# quick and dirty creation of a data frame, using a list of lists

import pandas as pd
from pandas import Series, DataFrame
```

```python
df = DataFrame([[10, 20, 30],     # list of lists
                [40, 50, 60],
                [70, 80, 90],
                [100, 110, 120]])
```

```python
df
```

- Rows are 0,1,2,3
- Columns are 0, 1, 2

This is because we didn't name them.

Notice also that it's OK for the rows and columns to have the same names (or in this case, numbers).




```python
df.index   # show me the rows
```

```python
df.columns  # show me the columns
```

```python
# I might want to set either the index or the column names
# we saw with a series that we can pass index= and then a list
# the same is true with a data frame, except we can also say columns= and give a list of column names

df = DataFrame([[10, 20, 30],     # list of lists
                [40, 50, 60],
                [70, 80, 90],
                [100, 110, 120]],
               index=list('abcd'),
               columns=list('xyz'))
df
```

```python
# just like with a series, we can retrieve via the index using .loc (and .iloc, by position)

df.loc['a']
```

```python
df.loc['c']
```

```python
# I can even use fancy indexing to get more than one row

df.loc[['a', 'c']]  # pass a list of indexes for fancy indexing
```

```python
# I can even use a slice

df.loc['a':'c']   # no inner list here, because it's a slice
```

```python
# I can still use my friend iloc, to go by location

df.iloc[0]
```

```python
df.iloc[[0, 2]]
```

We can retrieve from columns using `[]`. That is, if you use `.loc`, you get rows (via the index). But if you use `[]`, you get columns. 

This is why I told you last week not to use `[]` to retrieve rows, because you'll get confused.

```python
df['x']  # this returns the x column in df
```

```python
# can I use fancy indexing with columns? Yes!

df[['x', 'z']]   # list inside of []
```

```python

df = DataFrame([[10, 20.1, 30],     # list of lists
                [40, 50.2, 60],
                [70, 80.3, 90],
                [100, 110, 120]],
               index=list('abcd'),
               columns=list('xyz'))
df
```

# More complex queries

Let's say that I want to retrieve from my data frame based on a particular column's values -- but I don't want that column's values, rather I want to use one column as my mask index, and retrieve from another.

For example, I want the mean price of all items with a product_tax > 5.

The way to do this is with `.loc`, the 2-argument edition.

- The first argument to `.loc` (mandatory) is the row selector
    - one index value
    - a list of index values ("fancy indexing")
    - a slice
    - a boolean series for use as a "mask index"
- The second (optional) argument to `.loc` is a column selector
    - one column name
    - a list of column names

```python
df.loc[123, 'department']   # row selector = 123, column selector = 'department'
```

```python
# I'm going to spread these out, using "method chaining"

(
    df
    .loc[
        123,              # index, the row selector
        'department'      # column name, the column selector
    ]

)
```

```python
# I can ask for more than one row


(
    df
    .loc[
        [123, 456],        # index, the row selector -- fancy indexing
        'department'      # column name, the column selector
    ]

)
```

```python
# I can ask for more than one column

(
    df
    .loc[
        [123, 456],        # index, the row selector -- fancy indexing
        ['price', 'department']      # column name, the column selector
    ]

)
```

```python
# I can use a boolean series to retrieve certain rows
# get the mean price, where the product tax is > 5

(
    df
    .loc[
        df['product_tax'] > 5,        # index, the row selector -- fancy indexing
        'price'      # column name, the column selector
    ]
    .mean()  

)
```

# Methods and attributes

Remember that Pandas has a *huge* number of methods and attributes.

Most of the methods that you learned about with a series are *also* going to work on a data frame.

- If the series version returns a single value, then the data frame version will return a series (one value for each column)
- If the series version returns a series, then the data frame version will return a data frame (one series for each column)

```python
df = DataFrame([[10, 20, 30],     # list of lists
                [40, 50, 60],
                [70, 80, 90],
                [100, 110, 120]],
               index=list('abcd'),
               columns=list('xyz'))
df
```

```python
df['x'].mean()  # get column x, and calculate the mean
```

```python
df.mean()    #mean from each column
```

```python
df['y'].max()  # max from y
```

```python
df.max()  # the max value from each column
```

```python
df['z'].describe()   # this gives us back a series, describing the column
```

```python
df.describe()    # this gives us back a data frame, describing each of the columns
```

# Setting the index

One thing that we didn't really have with a series is the ability to set and remove the index. We could assign to the `.index` attribute, but it's not that same as a data frame.

In a data frame, we can move a column to be the index with the `set_index` method. We can push the index into being a regular column with the `reset_index` method.

**In neither case does the data frame really change!** The default in Pandas is to return a new value, based on the old one, with the change in place. But the change doesn't really happen on the original data frame.

```python
df
```

```python
df.set_index('x')  
```

```python
# have I really changed df? No!

df
```

```python
# if I want the set_index to be permanent, I assign the data frame I got back from set_index to df

df = df.set_index('x')
```

```python
df
```

```python
# to reverse this, and turn the index back into a regular (named) column, we use reset_index

df.reset_index()
```

```python
# just like set_index, reset_index doesn't really change anything -- it returns a new data frame

df = df.reset_index()
df
```

```python
df = df.set_index('y')
df
```

Pandas people use the index a lot, because a lot of methods use the index, and are dependent on it.

```python
df = df.reset_index()
```

```python
df
```

```python
# method chaining in action!

(
    df
    .set_index('x')
    .loc[[10, 70],   # row selector
         'y']        # column selector  -- this means, "give me column y where the row has index 10 or 70"
)
```

# What about `NaN`?

Remember `NaN`, the "not a number" value that indicates we don't have a value?

On a series:

- `dropna` returns a new series based on our existing one, but without the `NaN` values
- `fillna` returns a new series based on our existing one, but replacing all `NaN` values with something we specify

These methods, on data frames, work a bit differently:

- `dropna` returns a new data frame based on our existing one, but without any row containing a `NaN` value -- a single `NaN` value in the row will remove the whole row!
- `fillna` returns a new data frame based on our existing on. You can fill with a single value (which will fill all `NaN` values) or with a series whose index matches our columns -- and then each column's `NaN` can be filled in differently.


```python
df
```

```python
# add some nan values

import numpy as np
```

```python
df = DataFrame([[10, 20, 30],     # list of lists
                [40, 50, 60],
                [70, 80, 90],
                [100, 110, 120]],
               index=list('abcd'),
               columns=list('xyz'))


df.loc['b', 'y'] = np.nan
df.loc['a', 'z'] = np.nan
df.loc['a', 'x'] = np.nan
df.loc['d', 'y'] = np.nan

df
```

```python
df.dtypes
```

```python
df.dropna()  # what will we get back?
```

The good news with the above query is that we now have 100% rock-solid data. No NaNs!

The bad news is that we got rid of 75% of our data.

You can decide that some columns are more important than others, and should/shouldn't be included in the NaN decision. You can also decide that it's OK to have *some* NaN in a row, but maybe not everything.

```python
df.dropna(subset=['x'])  # only pay attention to NaN in x
```

```python
df.dropna(thresh=2)  # any row with at least 2 good values will be returned
```

# Boolean/mask indexes

Remember:

- We invoke, via broadcasting, an operator on a series
- That returns a series of booleans (`True` and `False`)
- We apply that boolean series to our series with `.loc`
- Only those values that correspond to `True` are returned -- the rest are dropped




```python
s = Series([10, 20, 30])

s.loc[[True, False, True]]
```

```python
s.loc[ s >= 20 ]
```

```python
df
```

```python
# we can use df.loc, giving it a boolean series as a mask index -- which it'll apply to every row

df.loc[    df['price'] > 30  ]
```

```python
# as of Pandas 3.0, we have a new, special syntax for this kind of thing, using pd.col(NAME) instead of df['NAME']

(
    df
    .loc[ pd.col('price') > 30 ]   # this is considered more modern
)
```

# Files!

When I started to work with data, I assumed there was a standard way to share data files in a robust way. Answer: NOPE! We still mostly use CSV (comma-separated values) and Excel files.

- CSV has one record per line, with commas (or something else) separating fields. This is a text file, which means that Pandas needs to guess (or we need to hint) regarding what types of values to expect.
- Excel is closer to binary in many ways, which can mean that we'll get values without guessing. However, Excel is often limited in how much data it can contain. Also, it can be very slow to load.

Pandas has facilities to both read and write 30 or so different formats. The two most common ones are still CSV and Excel.

To read a file into a Pandas data frame, you'll want to use a `pd.read_*` method. It's not a data-frame method, because we're getting a new data frame from it. Rather, it's a top-level `pd` method. 

```python
df = pd.read_csv('taxi.csv')
df
```
