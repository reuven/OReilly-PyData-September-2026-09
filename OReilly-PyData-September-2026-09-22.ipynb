# Data frames

If a series is a 1D data structure (with values and an index), then a data frame is a 2D data structure, much like an Excel spreadsheet. It also has an index (identifying the rows) and columns (identifying the ... columns).

Each column in a data frame is actually a series in disguise!

```python
# quick and dirty creation of a data frame, using a list of lists

import pandas as pd
from pandas import Series, DataFrame
```

```python
df = DataFrame([[10, 20, 30],     # list of lists
                [40, 50, 60],
                [70, 80, 90],
                [100, 110, 120]])
```

```python
df
```

- Rows are 0,1,2,3
- Columns are 0, 1, 2

This is because we didn't name them.

Notice also that it's OK for the rows and columns to have the same names (or in this case, numbers).




```python
df.index   # show me the rows
```

```python
df.columns  # show me the columns
```

```python
# I might want to set either the index or the column names
# we saw with a series that we can pass index= and then a list
# the same is true with a data frame, except we can also say columns= and give a list of column names

df = DataFrame([[10, 20, 30],     # list of lists
                [40, 50, 60],
                [70, 80, 90],
                [100, 110, 120]],
               index=list('abcd'),
               columns=list('xyz'))
df
```

```python
# just like with a series, we can retrieve via the index using .loc (and .iloc, by position)

df.loc['a']
```

```python
df.loc['c']
```

```python
# I can even use fancy indexing to get more than one row

df.loc[['a', 'c']]  # pass a list of indexes for fancy indexing
```

```python
# I can even use a slice

df.loc['a':'c']   # no inner list here, because it's a slice
```

```python
# I can still use my friend iloc, to go by location

df.iloc[0]
```

```python
df.iloc[2]
```

```python
df.iloc[[0, 2]]
```

```python
df.iloc[0:2]  # unlike .loc, this is up to and *not* including, as per Python in general
```

# But ... what about the columns?

We can retrieve from columns using `[]`. That is, if you use `.loc`, you get rows (via the index). But if you use `[]`, you get columns. 

This is why I told you last week not to use `[]` to retrieve rows, because you'll get confused.

```python
df['x']  # this returns the x column in df
```

```python
# can I use fancy indexing with columns? Yes!

df[['x', 'z']]   # list inside of []
```

```python

df = DataFrame([[10, 20.1, 30],     # list of lists
                [40, 50.2, 60],
                [70, 80.3, 90],
                [100, 110, 120]],
               index=list('abcd'),
               columns=list('xyz'))
df
```

```python
df.dtypes  # what dtypes do we have?
```

```python
# what happens if I now retrieve row a?

df.loc['a']   # we get a series back, with a dtype of float64 -- the integers were silently changed for this temporary, in-memory series
```

# Another way to retrieve columns

I much (MUCH) prefer to retrieve columns in this way, as I've shown you, with `[]`. But there is another way, namely dot notation. Just put the name of the column after the name of the data frame and a dot, as if it's an attribute of the data frame.

```python
df.price
```

```python
df.department
```

```python
df.product
```

Dot syntax works great, until/unless:

- Your column has the same name as a Pandas method or attribute, in which case the Pandas value gets priority over yours
- If the column contains spaces or other characters that aren't legal in Python attribute names, then they won't work

# More complex queries

Let's say that I want to retrieve from my data frame based on a particular column's values -- but I don't want that column's values, rather I want to use one column as my mask index, and retrieve from another.

For example, I want the mean price of all items with a product_tax > 5.

The way to do this is with `.loc`, the 2-argument edition.

- The first argument to `.loc` (mandatory) is the row selector
    - one index value
    - a list of index values ("fancy indexing")
    - a slice
    - a boolean series for use as a "mask index"
- The second (optional) argument to `.loc` is a column selector
    - one column name
    - a list of column names

```python
df.loc[123, 'department']   # row selector = 123, column selector = 'department'
```

```python
# I'm going to spread these out, using "method chaining"

(
    df
    .loc[
        123,              # index, the row selector
        'department'      # column name, the column selector
    ]

)
```

```python
# I can ask for more than one row


(
    df
    .loc[
        [123, 456],        # index, the row selector -- fancy indexing
        'department'      # column name, the column selector
    ]

)
```
